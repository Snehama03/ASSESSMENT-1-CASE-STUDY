#TASK8

```python
merged_df = pd.merge(
    population,
    gdp,
    on='country',
    how='inner'
)
```

```python
merged_df = pd.merge(
    merged_df,
    internet_users,
    on='country',
    how='inner'
)
```

```python
literacy_rate = pd.read_xml('/content/literacy_rate.xml')
literacy_rate.rename(columns={'name': 'country'}, inplace=True)

merged_df = pd.merge(
    merged_df,
    literacy_rate,
    on='country',
    how='inner'
)
```

```python
merged_df
```

```python
merged_df.columns
```

#TASK9

```python
merged_df['Internet Penetration (%)'] = (
    merged_df['internet_users'] / merged_df['population']
) * 100
```

```python
merged_df['GDP Per Capita'] = (
    merged_df['GDP'] / merged_df['population']
)
```

```python
merged_df['Internet Penetration (%)'].round(2)
```

```python
merged_df['GDP Per Capita'].round(2)
```

```python
merged_df[
        ['country', 'population', 'GDP', 'internet_users',
         'literacy_rate', 'Internet Penetration (%)', 'GDP Per Capita']
    ].head()
```

```python
# 11. Display the Top 5 countries by GDP:
q = "SELECT country, GDP FROM country_statistics ORDER BY GDP DESC LIMIT 5"
df = pd.read_sql_query(q, conn)
print(df)
print("\n")
```

```python
# 12. Find the average literacy rate:
q= "SELECT AVG(literacy_rate) FROM country_statistics"
df = pd.read_sql_query(q, conn)
print(df.iloc[0,0])
print("\n")
```

```python
# 13. Display all countries having literacy rate above 90%:
q= "SELECT country, literacy_rate FROM country_statistics WHERE literacy_rate > 90 ORDER BY literacy_rate DESC"
df= pd.read_sql_query(q, conn)
print(df)
print("\n")
```

```python
# 14. Find the Top 5 countries with the highest internet penetration:
q= "SELECT country, \"Internet Penetration (%)\" FROM country_statistics ORDER BY \"Internet Penetration (%)\" DESC LIMIT 5"
df = pd.read_sql_query(q, conn)
print(df)
print("\n")
```

```python
# 15. Calculate the total population:
q= "SELECT SUM(population) FROM country_statistics"
df = pd.read_sql_query(q, conn)
print(df.iloc[0,0])
print("\n")
```

```python
# 16. Find the country with the highest GDP per capita:
q = "SELECT country, \"GDP Per Capita\" FROM country_statistics ORDER BY \"GDP Per Capita\" DESC LIMIT 1"
df_highest_gdp_per_capita = pd.read_sql_query(q, conn)
print(df_highest_gdp_per_capita)
print("\n")
```

```python
# 17. Export the final merged dataset as a CSV file.

final_df = pd.read_sql_query("SELECT * FROM country_statistics", conn)
final_df.to_csv('country_statistics.csv', index=False)
print("Dataset exported to 'country_statistics.csv'.")
print("\n")
```

```python
# 18. Sort the dataset in descending order of GDP per Capita and display the Top 10 countries.
q = "SELECT country, \"GDP Per Capita\" FROM country_statistics ORDER BY \"GDP Per Capita\" DESC LIMIT 10"
df_top10_gdp_per_capita = pd.read_sql_query(q, conn)
print(df_top10_gdp_per_capita)
print("\n")
```
