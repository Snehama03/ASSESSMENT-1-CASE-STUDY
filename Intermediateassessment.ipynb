####Q1. Write a function that return the range (maximum and minimum) of a given list of numbers

```python
def range(num):
    return max(num),min(num)
num=[10,12,13,16,18]
print(range(num))
```

####Q3. Write a function that returns the median value of a list of numbers

```python
def median(num):
    num.sort()
    n = len(num)
    if n%2==1:
        return num[n//2]
    else:
        return (num[n//2-1]+num[n//2])/2
print(median([8,9,8,7,8]))
```

#####Q4. Write a function to return the mod of a list of numbers, if no mod exists, it should return None. If more than one mod exist, return the mod values as list

```python
def mode(numbers):
    counts = {}
    for num in numbers:
        counts[num] = counts.get(num, 0) + 1
    max_count = max(counts.values())
    if max_count == 1:
        return None
    modes = [num for num, count in counts.items() if count == max_count]
    if len(modes) == 1:
        return modes[0]
    else:
        return modes
print(mode([1,2,2,4,5]))
print(mode([1,2,3,3,5]))
print(mode([2,3,1,4,5]))
```

#####Q5. Write function to compute a given percentile value (e.g. 90th percentile) of a list of numbers

```python
def percentile(numbers, p):
    numbers.sort()
    n=len(numbers)
    position =(n - 1)*p/100
    lower=int(position)
    upper=lower + 1
    if upper >= n:
        return numbers[lower]
    fraction = position - lower
    return numbers[lower] + fraction * (numbers[upper] - numbers[lower])
print(percentile([10, 20, 30, 40, 50], 90))
```

#####Q7. Write function to return any outliers from a given list of numbers, if no outliers present it should return None

```python
def find_outliers(numbers):
    numbers.sort()
    n = len(numbers)
    q1 = numbers[n // 4]
    q3 = numbers[(3 * n) // 4]
    iqr = q3 - q1
    lower = q1 - 1.5 * iqr
    upper = q3 + 1.5 * iqr
    outliers = [num for num in numbers if num < lower or num > upper]
    return outliers if outliers else None
print(find_outliers([10, 12, 11, 13, 12, 14, 100]))
```

####Q8. Write a function to return standard deviation of a list of numbers

```python
import math
def standard_deviation(numbers):
    mean = sum(numbers) / len(numbers)
    variance = sum((x - mean) ** 2 for x in numbers) / len(numbers)
    return math.sqrt(variance)
print(standard_deviation ([20,20,30,40,40]))
```

####Q9. Write function to compute and return Z-score values of a list of numbers

```python
def z_scores(numbers):
    mean = sum(numbers) / len(numbers)
    variance = sum((x - mean) ** 2 for x in numbers) / len(numbers)
    std = variance**0.5
    return [(x - mean) / std for x in numbers]
print(z_scores([10, 20, 30, 40, 50]))
```

####Q10. Write a function to compute and return the Pearson correlation coefficient value of two list of numbers

```python
def pearson_correlation(x, y):
    mean_x = sum(x) / len(x)
    mean_y = sum(y) / len(y)
    numerator = sum((a - mean_x) * (b - mean_y) for a, b in zip(x, y))
    denominator = (
        sum((a - mean_x)**2 for a in x) *
        sum((b - mean_y)**2 for b in y)
    ) ** 0.5
    return numerator / denominator
print(pearson_correlation([1, 2, 3, 4, 5], [2, 4, 6, 8, 10]))
```
